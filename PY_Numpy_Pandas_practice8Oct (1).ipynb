```python
import numpy as np
import pandas as pd

# ==============================================================================
# SECTION A - EASY
# ==============================================================================

# --- NumPy ---

# 1. Array containing odd integers from 11 to 29
arr1 = np.arange(11, 30, 2)
print("Q1 Array:", arr1)
print("Q1 Dimensions:", arr1.ndim)
print("Q1 Total elements:", arr1.size)

# 2. Seven equally spaced values between 5 and 35
arr2 = np.linspace(5, 35, 7)
print("\nQ2 Equally spaced values:", arr2)

# 3. 4x3 array filled with 9, change element at (2, 1) to 0
arr3 = np.full((4, 3), 9)
arr3[2, 1] = 0  # 3rd row, 2nd column
print("\nQ3 Array:\n", arr3)

# 4. Calculate remainder when divided by 5
a4 = np.array([6, 12, 18, 24])
remainder = a4 % 5
print("\nQ4 Remainder:", remainder)

# 5. Extract elements at indices 1, 3, and 5 using slicing
arr5 = np.array([14, 28, 42, 56, 70, 84])
extracted = arr5[1::2]
print("\nQ5 Extracted elements:", extracted)

# 6. Absolute value and square root
arr6 = np.array([-16, 25, -36, 49])
abs_val = np.abs(arr6)
sq_root = np.sqrt(abs_val)
print("\nQ6 Absolute values:", abs_val)
print("Q6 Square roots:", sq_root)

# 7. 3x2 matrix transpose and shapes
matrix7 = np.arange(7, 13).reshape(3, 2)
transpose7 = matrix7.T
print("\nQ7 Shape before transpose:", matrix7.shape)
print("Q7 Transpose:\n", transpose7)
print("Q7 Shape after transpose:", transpose7.shape)

# 8. Range and median temperature
temperatures = np.array([18, 22, 27, 31, 24, 19])
temp_range = np.max(temperatures) - np.min(temperatures)
temp_median = np.median(temperatures)
print("\nQ8 Temperature Range:", temp_range)
print("Q8 Temperature Median:", temp_median)

# 9. Distinct values in ascending order and count
values = np.array([4, 9, 4, 2, 9, 7, 2, 4])
distinct_values = np.unique(values)
print("\nQ9 Distinct values:", distinct_values)
print("Q9 Count of distinct values:", len(distinct_values))

# 10. Speed classification using np.where()
speed = np.array([35, 62, 48, 81, 55])
category = np.where(speed >= 60, "Fast", "Normal")
print("\nQ10 Speed Categories:", category)

# --- Pandas Series ---

# 11. Books issued series
books = pd.Series(
    [24, 31, 19, 27], index=["Monday", "Tuesday", "Wednesday", "Thursday"]
)
print("\nQ11 Series:\n", books)

# 12. Retrieve Tablet price
prices = pd.Series([45000, 18000, 22000], index=["Laptop", "Tablet", "Phone"])
print("\nQ12 Tablet Price:", prices["Tablet"])

# 13. Filter rainfall measurements > 10
rainfall = pd.Series([12, 0, 18, 5, 23])
print("\nQ13 Rainfall > 10:\n", rainfall[rainfall > 10])

# 14. Series from dictionary (index and values)
rates_dict = {"Gold": 7200, "Silver": 95, "Copper": 850}
rates_series = pd.Series(rates_dict)
print("\nQ14 Index:", rates_series.index)
print("Q14 Values:", rates_series.values)

# 15. Assign Series name and index name
ratings = pd.Series([3.5, 4.2, 2.8, 4.7])
ratings.name = "Customer Ratings"
ratings.index.name = "Review ID"
print("\nQ15 Series:\n", ratings)

# 16. Check label existence
s16 = pd.Series([11, 22, 33], index=["X1", "X2", "X3"])
print("\nQ16 Is 'X2' in index?", "X2" in s16.index)
print("Q16 Is 'X4' absent from index?", "X4" not in s16.index)

# 17. Convert Series to dictionary
stock = pd.Series([16, 28, 12], index=["Keyboard", "Mouse", "Monitor"])
print("\nQ17 Dictionary:", stock.to_dict())

# 18. Dictionary with reordered index
data_dict18 = {"Hindi": 78, "English": 84, "Computer": 91}
s18 = pd.Series(data_dict18, index=["Computer", "Hindi", "Art", "English"])
print("\nQ18 Series:\n", s18)
# Missing entry: 'Art' is NaN because it is not in the source dictionary.

# 19. isnull() and notnull()
expenses = pd.Series([1200, None, 850, None, 600])
print("\nQ19 isnull():\n", expenses.isnull())
print("\nQ19 notnull():\n", expenses.notnull())

# 20. Addition with aligned index labels
a20 = pd.Series([8, 15, 22], index=["P", "Q", "R"])
b20 = pd.Series([4, 6, 10], index=["R", "P", "Q"])
print("\nQ20 Series Addition:\n", a20 + b20)


# ==============================================================================
# SECTION B - MODERATE
# ==============================================================================

# --- NumPy ---

# 21. Sales calculations
sales = np.array(
    [[120, 150, 180], [140, 160, 175], [135, 155, 190], [150, 170, 200]]
)
daily_sales = sales.sum(axis=1)
product_sales = sales.sum(axis=0)
best_day = np.argmax(daily_sales) + 1
print("\nQ21 Total daily sales:", daily_sales)
print("Q21 Total product sales:", product_sales)
print(f"Q21 Day with highest sales: Day {best_day}")

# 22. Submatrix mean
arr_6x6 = np.arange(10, 46).reshape(6, 6)
submatrix = arr_6x6[2:6, 1:5]
print("\nQ22 Submatrix:\n", submatrix)
print("Q22 Submatrix Mean:", submatrix.mean())

# 23. Boolean indexing and replacing values
weights = np.array([48, 62, 75, 81, 54, 69, 90])
weights_above_70 = weights[weights > 70]
new_weights = np.where(weights < 55, 55, weights)
print("\nQ23 Weights above 70:", weights_above_70)
print("Q23 Modified weights:", new_weights)
print("Q23 Original weights (unchanged):", weights)

# 24. Bill calculation and contribution percentage
quantity = np.array([3, 5, 2, 7])
price = np.array([120, 80, 250, 60])
total_cost = quantity * price
overall_bill = np.sum(total_cost)
contrib_pct = (total_cost / overall_bill) * 100
print("\nQ24 Total cost per product:", total_cost)
print("Q24 Overall bill:", overall_bill)
print("Q24 Contribution (%):", np.round(contrib_pct, 2))

# 25. Variance and Standard Deviation (axis=0 vs axis=1)
arr25 = np.array([[2, 4, 6, 8], [1, 3, 5, 7], [10, 20, 30, 40]])
print("\nQ25 Row Variance (axis=1):", np.var(arr25, axis=1))
print("Q25 Row Std Dev (axis=1):", np.std(arr25, axis=1))
print("Q25 Column Variance (axis=0):", np.var(arr25, axis=0))
print("Q25 Column Std Dev (axis=0):", np.std(arr25, axis=0))

# 26. Matrix multiplication
A = np.array([[1, 2, 3], [4, 5, 6]])  # 2x3
B = np.array([[7, 8], [9, 10], [11, 12]])  # 3x2
print("\nQ26 Matrix Multiplication (2x3 @ 3x2):\n", np.matmul(A, B))

# 27. Random integers, sorting, and unique elements
np.random.seed(42)
random_vals = np.random.randint(10, 100, size=25)
sorted_vals = np.sort(random_vals)
print("\nQ27 Sorted Values:", sorted_vals)
print("Q27 Five Largest Values:", sorted_vals[-5:])
print("Q27 Unique Count:", len(np.unique(random_vals)))

# 28. View vs Independent Copy
a28 = np.array([5, 10, 15, 20, 25, 30])
b28 = a28[1:5]
c28 = a28[1:5].copy()
b28[1] = 99
c28[2] = 77
print("\nQ28 Final array 'a':", a28)
print("Q28 Final array 'b' (view):", b28)
print("Q28 Final array 'c' (copy):", c28)

# --- Pandas Series ---

# 29. Monthly revenue addition
dept_a = pd.Series({"Jan": 250, "Feb": 310, "Mar": 290})
dept_b = pd.Series({"Mar": 180, "Jan": 210, "Apr": 260})
print("\nQ29 Total Revenue:\n", dept_a + dept_b)

# 30. Filter attendance < 75%
attendance = pd.Series(
    [82, 67, 91, 74, 58], index=["S101", "S102", "S103", "S104", "S105"]
)
low_att = attendance[attendance < 75]
print("\nQ30 Attendance < 75%:\n", low_att)
print("Q30 Student Labels:", low_att.index.tolist())
print("Q30 Count:", len(low_att))

# 31. Inventory Series with explicit labels
inventory = {"Pen": 50, "Pencil": 80, "Eraser": 40, "Scale": 25}
labels = ["Scale", "Pen", "Marker", "Eraser"]
s31 = pd.Series(inventory, index=labels)
print("\nQ31 Inventory Series:\n", s31)

# 32. Label-based Series subtraction
s32_1 = pd.Series([100, 200, 300], index=["North", "South", "East"])
s32_2 = pd.Series([10, 20, 30], index=["South", "West", "North"])
res32 = s32_1 - s32_2
print("\nQ32 Subtraction Result:\n", res32)
print("Q32 Non-missing entries:\n", res32[res32.notnull()])

# 33. Filter, dictionary conversion, and reconstruction
electricity = pd.Series(
    [210, 310, 180, 290, 400],
    index=["House_A", "House_B", "House_C", "House_D", "House_E"],
)
filtered_elec = electricity[electricity > 250]
reconstructed_elec = pd.Series(filtered_elec.to_dict())
print("\nQ33 Reconstructed Electricity Series:\n", reconstructed_elec)

# 34. Filtering, missing values, and metadata
s34 = pd.Series({"A": 45, "B": 70, "C": 55}, index=["C", "D", "A", "B"])
s34.name = "Performance"
s34.index.name = "Group"
print("\nQ34 s >= 50:\n", s34[s34 >= 50])
print("\nQ34 Missing entries (isnull):\n", s34[s34.isnull()])
print("Q34 Series Name:", s34.name)
print("Q34 Index Name:", s34.index.name)
```
