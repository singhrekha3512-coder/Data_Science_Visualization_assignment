```python
course_df = pd.DataFrame(
    {
        "Student": ["Alice", "Bob", "Charlie", "David", "Eva"],
        "Course": ["CS", "AI", "DS", "AI", "CS"],
        "Score": [88, 92, 79, 85, 90],
    }
)

print("\nQ26 - Step 1 (Column Names):", course_df.columns)
print("Q26 - Step 2 (Retrieve Student):\n", course_df["Student"])

student_score_df = course_df[["Student", "Score"]]
print("Q26 - Step 3 (Student & Score DataFrame):\n", student_score_df)

course_df["Batch"] = "2026"
print("Q26 - Step 4 (Added Batch Column):\n", course_df)
```

```python
travel_df = pd.DataFrame(
    {
        "Destination": ["Paris", "Tokyo", "London", "New York", "Rome"],
        "Days": [5, 7, 4, 6, 5],
    }
)
# Step A: Add Code using np.arange(5)
travel_df["Code"] = np.arange(5)
print("\nQ27 - After adding Code column:\n", travel_df)

# Step B: Replace Code column with Series indexed at 1 and 4
code_series = pd.Series(["TK-01", "RM-05"], index=[1, 4])
travel_df["Code"] = code_series
print("\nQ27 - After replacing Code column with partially indexed Series:\n", travel_df)
```

```python
school_df = pd.DataFrame(
    {
        "Student": ["Aarav", "Vivaan", "Aditi"],
        "Class": [10, 10, 10],
        "Section": ["A", "B", "A"],
    }
)

school_df_house = pd.DataFrame(
    school_df, columns=["Student", "Class", "Section", "House"]
)
print("\nQ28 - Before assigning House value:\n", school_df_house)

school_df_house["House"] = "Red"
print("\nQ28 - After assigning common House value:\n", school_df_house)
```

```python
inventory_df = pd.DataFrame(
    {
        "Item": ["Laptop", "Mouse", "Keyboard", "Monitor"],
        "Qty": [10, 50, 30, 15],
    }
)

# Add column using partially indexed Series (only indexes 0 and 2)
part_series = pd.Series(["A1", "C3"], index=[0, 2])
inventory_df["Aisle"] = part_series
print("\nQ29 - DataFrame with partially indexed column:\n", inventory_df)

# Delete added column
del inventory_df["Aisle"]
print("\nQ29 - DataFrame after deleting column:\n", inventory_df)
```
